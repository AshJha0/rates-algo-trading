# Trade size by tenor and clustering (EDGE-E4)

Censored lognormal size distributions per tenor (2, 5, 10, 30 years) of USD SOFR OIS prints from DTCC SDR, the out-of-sample calibration of the fitted 95th percentile, and the excess of same-tenor trades in the 5 minutes after a large trade. Pre-registration `research/prereg/E4.yaml`; fitting 2025-01-02 to 2025-06-30, holdout 2025-07-01 to 2026-06-30. FINRA Treasury trades need an account; the FINRA fixture is synthetic and enters no test.

Every number below is recomputed from the committed fixtures and checked against the ledger row; the plain-language finding is in `docs/research/EDGE_STUDIES.md`.

```python
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

from ratesresearch import backtest_analysis as ba
from ratesresearch.edge import report

ROOT = Path.cwd()
while not (ROOT / "data" / "presets.yaml").is_file():
    ROOT = ROOT.parent


def show(df: pd.DataFrame, digits: int = 4) -> None:
    display(Markdown(ba.to_markdown(df.round(digits))))
```

```python
from ratesresearch.edge import e4_size

res = e4_size.run(ROOT)
row = report.ledger_row(ROOT, "EDGE-E4")
display(Markdown(report.check_against_ledger(res, row)))
show(report.tests_table(res, row))
```
