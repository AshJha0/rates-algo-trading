# Spread optimisation

Thin wrapper over `ratesresearch.spread`: optimal half-spread and expected edge per product across the declared band, and the sensitivity to volatility and dealer count. Adverse selection and hedge cost are the fitting-period estimates pinned in `data/experiments/P5-001.config.yaml` (tier 2 adverse, 0.0667 bp; hedge cost, 0.2459 bp).

```python
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

from ratesresearch import backtest_analysis as ba
from ratesresearch import presets

ROOT = Path.cwd()
while not (ROOT / "data" / "presets.yaml").is_file():
    ROOT = ROOT.parent
PRESETS = ROOT / "data" / "presets.yaml"


def show(df: pd.DataFrame) -> None:
    display(Markdown(ba.to_markdown(df)))
```

```python
from ratesresearch import spread

hit = presets.get_preset(PRESETS, "usd_rfq_hit_rate")
comp = presets.get_preset(PRESETS, "usd_rfq_competition")
show(pd.DataFrame(spread.band_table(hit, comp, adverse_bp=0.0667, hedge_bp=0.2459)))
show(pd.DataFrame(spread.sensitivity(hit, comp, "SOFR 10Y", 0.0667, 0.2459)))
```
