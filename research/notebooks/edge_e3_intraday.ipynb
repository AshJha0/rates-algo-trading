# Intraday volatility and intensity around releases (EDGE-E3)

Realised variance and trade counts of USD SOFR OIS prints (DTCC SDR public dissemination) per 5-minute bucket, on 08:30 ET release days, FOMC days and around the 08:00 ET SOFR publication, against control days. Pre-registration `research/prereg/E3.yaml`; fitting 2025-01-02 to 2025-06-30, holdout 2025-07-01 to 2026-06-30. Prints are executions across the market, not one dealer's RFQs.

Every number below is recomputed from the committed fixtures and checked against the ledger row; the plain-language finding is in `docs/research/EDGE_STUDIES.md`.

```python
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

from ratesresearch import backtest_analysis as ba
from ratesresearch.edge import report

ROOT = Path.cwd()
while not (ROOT / "data" / "presets.yaml").is_file():
    ROOT = ROOT.parent


def show(df: pd.DataFrame, digits: int = 4) -> None:
    display(Markdown(ba.to_markdown(df.round(digits))))
```

```python
from ratesresearch.edge import e3_intraday

res = e3_intraday.run(ROOT)
row = report.ledger_row(ROOT, "EDGE-E3")
display(Markdown(report.check_against_ledger(res, row)))
show(report.tests_table(res, row))
```
