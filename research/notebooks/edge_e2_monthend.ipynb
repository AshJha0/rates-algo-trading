# Month-end index extension (EDGE-E2)

Does 2s30s flatten over the last 3 business days of months with large coupon issuance (term-weighted, top third of the 2015 to 2020 distribution) by more than an ordinary 3-day window? Pre-registration `research/prereg/E2.yaml`; holdout 2021-01 to 2026-06. The index extension itself is proprietary; the issuance proxy stands in for it.

Every number below is recomputed from the committed fixtures and checked against the ledger row; the plain-language finding is in `docs/research/EDGE_STUDIES.md`.

```python
from pathlib import Path

import pandas as pd
from IPython.display import Markdown, display

from ratesresearch import backtest_analysis as ba
from ratesresearch.edge import report

ROOT = Path.cwd()
while not (ROOT / "data" / "presets.yaml").is_file():
    ROOT = ROOT.parent


def show(df: pd.DataFrame, digits: int = 4) -> None:
    display(Markdown(ba.to_markdown(df.round(digits))))
```

```python
from ratesresearch.edge import e2_monthend

res = e2_monthend.run(ROOT)
row = report.ledger_row(ROOT, "EDGE-E2")
display(Markdown(report.check_against_ledger(res, row)))
show(report.tests_table(res, row))
```
